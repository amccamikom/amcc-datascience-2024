#Import Library

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#Load Dataset

```python
df1 = pd.read_csv("https://raw.githubusercontent.com/RachmasariAR/CostOfLiving/refs/heads/main/Cost_of_Living_2018.csv")
df2 = pd.read_csv("https://raw.githubusercontent.com/RachmasariAR/CostOfLiving/refs/heads/main/Cost_of_Living_2022.csv")
```

```python
df2
```

```python
df2 = df2.drop("Rank", axis=1)
```

#Model Linear Regression

```python
from sklearn.model_selection import train_test_split
x = df.drop(['Cost of Living Plus Rent Index', 'Cost of Living Index'], axis=1)
y = df['Cost of Living Plus Rent Index']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)

print("Total sampel dari seluruh dataset : ", (len(x)))
print("Total sampel dari dataset train : ", (len(x_train)))
print("Total sampel dari dataset test : ", (len(x_test)))
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()

model.fit(x_train, y_train)

y_train_pred = model.predict(x_train)
y_test_pred = model.predict(x_test)
```

#Testing

```python
new_data= {
    'Rent Index' : [36.84],
    'Groceries Index' : [77.44],
    'Restaurant Price Index' : [72.95],
    'Local Purchasing Power Index' : [104.63]
}

new_df = pd.DataFrame(new_data)

predicted_cost = model.predict(new_df)
print("Hasil Prediksi dari Data di atas adalah : ", predicted_cost)
```

#Evaluasi Model

```python
from sklearn.metrics import r2_score
r2_score(y_test, y_test_pred)
```
